```python
#Count technologies by number of jobs
# Example:
#{
    #'Python': 7,
    #'PyTorch': 6,
    #'TensorFlow': 5,
#}
```

```python
tech_jobs_count = {}

for jobtech in jobs_df['technologies']:

    if jobtech is not None:

        for tech in set(jobtech):
            tech = tech.lower() #This removes the similarities like Python, python, PYTHON

            if tech not in tech_jobs_count:
                tech_jobs_count[tech] = 1

            else:
                tech_jobs_count[tech] += 1
tech_jobs_count
```

```python
techjob_series = pd.Series(tech_jobs_count)
techjob_series = techjob_series.sort_values(ascending=False)
techjob_series
```

```python
technologies_percentage = techjob_series / len(jobs_df) * 100
for t,p in technology_percentages.items():
    print(f"{t} appears in {p:.1f}% of the job postings.")
```

```python
skill_job_counts = {}

for skilljob in jobs_df['skills']:
    if skilljob is not None:

        for skill in set(skilljob):

            skill = skill.lower()

            if skill not in skill_job_counts:
                skill_job_counts[skill] = 1
            else:
                skill_job_counts[skill] += 1
```

```python
# Let's see which technologies are appearing together with Python
python_jobs = []

for technologies in jobs_df['technologies']:
    if technologies is not None:
        tech_lower = [tech.lower() for tech in technologies]

        if 'python' in tech_lower:
            python_jobs.append(tech_lower)

python_jobs
```

```python
python_technologies = []

for job in python_jobs:
    for tech in set(job):
        if tech != 'python':
            python_technologies.append(tech)
python_technologies
```

```python
python_series = pd.Series(python_technologies)
python_series.value_counts().head(10)
```
